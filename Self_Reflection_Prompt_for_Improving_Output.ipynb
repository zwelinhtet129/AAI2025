```python
# Exercise 3: Self-Reflection Prompt for Improving Output
# Objective: Ask the AI to critique and improve its own summary
# against clear requirements (self-reflection / critique-and-revise).
# Scenario: Summarize a short article about sleep for a busy college student.


ARTICLE = """
Sleep is essential for both physical and mental health. During sleep, the brain
consolidates memories, meaning that things you learned during the day are
strengthened overnight. Lack of sleep is linked to reduced concentration, weaker
immune function, and higher stress levels. Studies show that college students
who sleep 7 to 9 hours per night tend to earn better grades than those who
regularly sleep less than 6 hours. Experts recommend keeping a consistent sleep
schedule, avoiding screens before bed, and limiting caffeine in the afternoon.
"""


def summary_agent(prompt):
    """
    Simulates an AI agent that writes and revises summaries.
    In a real application, this would make an API call to a large language model.
    """
    print(f"--- Sending Prompt to AI Agent ---\n{prompt}\n----------------------------------")
    if "critique" in prompt.lower():
        response = """
Critique of my first summary:
- Accuracy: correct, nothing false.
- Clarity: weak, it rambles and buries the main point.
- Audience: too formal for a busy college student.
- Length: too long, 6 sentences is more than needed.
- Formatting: a plain paragraph, hard to scan quickly.

Revised summary:
- Sleep 7-9 hours a night to boost memory, focus, and grades.
- Too little sleep hurts concentration and immunity.
- Keep a regular schedule and avoid screens before bed.
"""
    else:
        response = """
Sleep is very important for physical and mental health because during sleep the
brain consolidates memories which means things learned during the day are
strengthened overnight, and a lack of sleep is linked to reduced concentration
and weaker immune function as well as higher stress levels, and studies show
college students sleeping 7 to 9 hours earn better grades than those sleeping
less than 6 hours, so experts recommend a consistent schedule, no screens
before bed, and less caffeine in the afternoon.
"""
    return response


# --- Iteration 1: the first summary ---
initial_prompt = f"Summarize this article:\n{ARTICLE}"
print("--- Response from Initial Prompt ---")
first_summary = summary_agent(initial_prompt)
print(first_summary)
print("Analysis: The summary is accurate but too long, unstructured, and not "
      "written for a busy student. It needs a self-review.\n\n")

# --- Iteration 2: self-reflection with explicit criteria ---
reflection_prompt = f"""
You are a careful editor. Critique the summary you just wrote against these
criteria, then provide a revised version:
1. Accuracy: is every claim supported by the article?
2. Clarity: is the main point easy to find?
3. Audience: is it written for a busy college student?
4. Length: no more than 3 short points.
5. Formatting: use bullet points so it is easy to scan.

Your summary was:
{first_summary}
"""
print("--- Response from Self-Reflection Prompt ---")
print(summary_agent(reflection_prompt))
print("Analysis: Success! The critique found real problems (length, clarity, "
      "formatting) and the revised summary is short, clear, and easy to scan.")
```
