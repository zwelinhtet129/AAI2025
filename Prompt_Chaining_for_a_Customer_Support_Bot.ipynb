```python
# Exercise 1: Prompt Chaining for a Customer Support Bot
# Objective: Build a multi-step flow where the output of one prompt
# becomes the input to the next (a "prompt chain").
# Scenario: An e-commerce store that sells electronics.
# A customer reports a defective laptop bought 15 days ago.


def customer_support_bot(prompt):
    """
    Simulates an AI agent responding to a customer query.
    In a real application, this function would make an API call
    to a large language model.
    """
    print(f"--- Sending Prompt to AI Agent ---\n{prompt}\n----------------------------------")
    p = prompt.lower()
    if "classify" in p:
        response = '{"category": "product", "urgency": "medium"}'
    elif "escalat" in p:
        response = "Decision: RESOLVE. Reason: standard defective-item case within policy."
    elif "propose" in p:
        response = ("1) Look up the order. 2) Start a replacement under the "
                    "30-day defective-item policy. 3) Send a confirmation email.")
    elif "ask only" in p:
        # improved gather prompt: asks only for what is missing
        response = "Could you please share your order number?"
    elif "missing details" in p:
        # initial gather prompt: no rules, so it asks for everything
        response = ("I need your order number, product name, purchase date, "
                    "and a full description of the problem.")
    else:
        response = "No simulated response for this prompt."
    return response


# The customer ticket that starts the chain
ticket = "My laptop arrived with a cracked screen. I bought it 15 days ago."

# --- Step 1: Classify the issue ---
print("=== Step 1: Classify ===")
classify_prompt = """
You are a support triage assistant for 'ElectroGadget Store'.
Classify the customer's issue.
Customer message: "My laptop arrived with a cracked screen. I bought it 15 days ago."
Reply with ONLY JSON: {"category": one of billing/shipping/product/other,
"urgency": low/medium/high}. No extra text.
"""
classification = customer_support_bot(classify_prompt)
print("AI Response:", classification)

# --- Step 2: Gather missing info (initial vs improved, like the class example) ---
print("\n=== Step 2: Gather missing info ===")
print("--- Initial, less effective prompt ---")
initial_gather = f"""
You are a polite support agent.
Customer message: "{ticket}"
Classification: {classification}
Ask the customer for missing details.
"""
print("AI Response:", customer_support_bot(initial_gather))

print("\n--- Improved prompt ---")
improved_gather = f"""
You are a polite and friendly support representative for 'ElectroGadget Store'.
Customer message: "{ticket}"
Classification: {classification}
Look at what the customer already told us and ask ONLY for the details that
are still missing (order number, product name, or purchase date).
Keep your reply to 2 sentences max.
"""
gathered = customer_support_bot(improved_gather)
print("AI Response:", gathered)

# --- Step 3: Propose a solution (uses Step 1 and Step 2 outputs) ---
print("\n=== Step 3: Propose a solution ===")
propose_prompt = f"""
You are a helpful support representative for 'ElectroGadget Store'.
Classification: {classification}
Information collected: {gathered}
Propose a solution as a numbered list of at most 3 concrete steps.
Stay within store policy: defective items get a replacement or full refund
within 30 days of purchase.
"""
solution = customer_support_bot(propose_prompt)
print("AI Response:", solution)

# --- Step 4: Escalation rule (uses Step 1 and Step 3 outputs) ---
print("\n=== Step 4: Escalation rule ===")
escalate_prompt = f"""
You are a support policy checker for 'ElectroGadget Store'.
Classification: {classification}
Proposed solution: {solution}
Decide whether this ticket should be resolved automatically or escalated
to a human supervisor. ESCALATE if urgency is high or the customer mentions
legal action. Otherwise choose RESOLVE.
Reply exactly: Decision: <ESCALATE or RESOLVE>. Reason: <one sentence>.
"""
print("AI Response:", customer_support_bot(escalate_prompt))
```
